### Print 1 to N

```python
def printnum(current, n):
    if current > n:
        return
    print(current, end=" ")
    printnum(current + 1, n)


def main():
    n = int(input("Enter n: "))
    printnum(1, n)


if __name__ == "__main__":
    main()
```

### Print N to 1

```python
def printnum(current):
    if current < 1:
        return
    print(current, end=" ")
    printnum(current - 1)


def main():
    n = int(input("Enter n: "))
    printnum(n)


if __name__ == "__main__":
    main()
```

### Sum of N natural no.

```python
def sumN(n):
    if n == 1:
        return 1
    return n + sumN(n - 1)


def main():
    n = int(input("Enter n: "))
    print(f"Sum of first {n} no.: {sumN(n)}")


if __name__ == "__main__":
    main()
```

### Factorial


```python
def fact(n):
    if n == 0:
        return 1
    return n * fact(n - 1)


def main():
    n = int(input("Enter a no.: "))
    print(f"Factorial of {n}: {fact(n)}")


if __name__ == "__main__":
    main()
```

### Reverse an array

```python
def revarr(arr):
    arr = arr[::-1]
    return arr


def main():
    arr = [1, 2, 3, 4, 5]
    res = revarr(arr)
    print(res)


if __name__ == "__main__":
    main()
```

### Fibonacci

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)


def main():
    n = int(input("Enter a no.: "))
    print(f"Fibonacci of {n}: {fib(n)}")


if __name__ == "__main__":
    main()
```
