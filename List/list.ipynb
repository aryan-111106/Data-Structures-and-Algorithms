### Average or mean of a list

```python
def average(l):
    sum = 0
    for i in l:
        sum = sum + i
    n = len(l)
    return sum / n


l = [10, 20, 30, 40]
print("Average of list: ", average(l))
```

```python
def average(l):
    return sum(l) / len(l)


l = [10, 20, 30, 40]
print("Average of list: ", average(l))
```

### Separate even and odd

```python
def separate(l):
    even = []
    odd = []
    for i in l:
        if i % 2 == 0:
            even.append(i)
        else:
            odd.append(i)
    return even, odd


l = [10, 41, 30, 15, 80]
even, odd = separate(l)
print("Even: ", even)
print("Odd: ", odd)
```

### Get smaller elements

```python
def small(l, x):
    s = []
    for i in l:
        if i < x:
            s.append(i)
    return s


l = [8, 100, 20, 40, 3, 7]
x = 10
print(small(l, x))
```

### Slicing(List, Tuple, String)

```python
l = [10, 20, 30, 40, 50]
print(l[0:5:2])
print(l[:4])
print(l[2:])
print(l[1:4])
print(l[4:1:-1])
print(l[-1:-6:-1])
print(l[::-1])
print(l[0:5])
print(l[:])

l1 = [10, 20, 30]
l2 = l1[:]

t1 = (10, 20, 30)
t2 = t1[:]

s1 = "geeks"
s2 = s1[:]

print(l1 is l2)
print(t1 is t2)
print(s1 is s2)
```

### Comprehension in Python

```python
def getSmaller(l, x):
    return [i for i in l if i < x]


l = [9, 15, 12, 3, 7, 11]
x = 10
print(getSmaller(l, x))
```
