### Average or mean of a list

```python
def average(l):
    sum = 0
    for i in l:
        sum = sum + i
    n = len(l)
    return sum / n


l = [10, 20, 30, 40]
print("Average of list: ", average(l))
```

```python
def average(l):
    return sum(l) / len(l)


l = [10, 20, 30, 40]
print("Average of list: ", average(l))
```

### Separate even and odd

```python
def separate(l):
    even = []
    odd = []
    for i in l:
        if i % 2 == 0:
            even.append(i)
        else:
            odd.append(i)
    return even, odd


l = [10, 41, 30, 15, 80]
even, odd = separate(l)
print("Even: ", even)
print("Odd: ", odd)
```

### Get smaller elements

```python
def small(l, x):
    s = []
    for i in l:
        if i < x:
            s.append(i)
    return s


l = [8, 100, 20, 40, 3, 7]
x = 10
print(small(l, x))
```

### Comprehension in Python

```python
def getSmaller(l, x):
    return [i for i in l if i < x]


l = [9, 15, 12, 3, 7, 11]
x = 10
print(getSmaller(l, x))
```

```python
def getEvenOdd(l):
    even = [x for x in l if x % 2 == 0]
    odd = [x for x in l if x % 2 != 0]
    return even, odd


l = [10, 3, 20, 5, 12]
even, odd = getEvenOdd(l)
print(even)
print(odd)
```

### Largest element in a list

```python
def getLargest(l):
    if not l:
        return None
    else:
        big = l[0]
        for i in range(1, len(l)):
            if l[i] > big:
                big = l[i]
        return big


l = [10, 5, 20, 8]
print(getLargest(l))
```

```python
def getLargest(l):
    for x in l:
        for y in l:
            if y > x:
                break
        else:
            return x
    return None


l = [10, 5, 20, 8]
print(getLargest(l))
```

### Second largest element in a list

```python
def getLargest(l):
    if not l:
        return None
    else:
        big = l[0]
        for i in range(1, len(l)):
            if l[i] > big:
                big = l[i]
        return big


def getsecLargest(l):
    if len(l) <= 1:
        return None
    lar = getLargest(l)
    slar = None
    for x in l:
        if x != lar:
            if slar == None:
                slar = x
            else:
                slar = max(slar, x)
    return slar


l = [5, 20, 12, 10, 20, 10, 12]
print(getsecLargest(l))
```

```python
def getsecLargest(l):
    if len(l) <= 1:
        return None
    lar = l[0]
    slar = None
    for x in l[1:]:
        if x > lar:
            slar = lar
            lar = x
        elif x != lar:
            if slar == None or slar < x:
                slar = x
    return slar


l = [5, 20, 12, 10, 20, 10, 12]
print(getsecLargest(l))
```

### Check if a list is sorted

```python
def sorted(l):
    sorted = 1
    for i in range(len(l) - 1):
        for j in range(len(l) - 1 - i):
            if l[j] > l[j + 1]:
                sorted = 0
    if sorted:
        return True
    else:
        return False


l = [1, 2, 3, 4]
print(sorted(l))
```

```python
def isSorted(l):
    i = 1
    while i < len(l):
        if l[i] < l[i - 1]:
            return False
        i = i + 1
    return True


l = [10, 20, 30, 15, 40]

if isSorted(l):
    print("Yes")
else:
    print("No")
```

```python
def isSorted(l):
    sl = sorted(l)
    if sl == l:
        return True
    else:
        return False


l = [10, 20, 30, 15, 40]
if isSorted(l):
    print("Yes")
else:
    print("No")
```

### Find the only odd

```python
def odd(l):
    res = None
    for i in l:
        count = l.count(i)
        if count % 2 != 0:
            res = i
            break
    return res


l = [10, 30, 30, 10, 30, 30, 20]
print(odd(l))
```

```python
def odd(l):
    res = 0
    for x in l:
        res = res ^ x
    return res


l = [10, 30, 30, 10, 30, 30, 20]
print(odd(l))
```

### Reverse a list

```python
def rev(l):
    rl = []
    for i in l:
        rl.insert(0, i)
    return rl


l = [10, 20, 30, 40]
print(rev(l))
```

```python
def revlist(l):
    s = 0
    e = len(l) - 1
    while s < e:
        l[s], l[e] = l[e], l[s]
        s = s + 1
        e = e - 1


l = [10, 20, 30, 40]
revlist(l)
print(l)
```
