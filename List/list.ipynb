### Average or mean of a list

```python
def average(l):
    sum = 0
    for i in l:
        sum = sum + i
    n = len(l)
    return sum / n


l = [10, 20, 30, 40]
print("Average of list: ", average(l))
```

```python
def average(l):
    return sum(l) / len(l)


l = [10, 20, 30, 40]
print("Average of list: ", average(l))
```

### Separate even and odd

```python
def separate(l):
    even = []
    odd = []
    for i in l:
        if i % 2 == 0:
            even.append(i)
        else:
            odd.append(i)
    return even, odd


l = [10, 41, 30, 15, 80]
even, odd = separate(l)
print("Even: ", even)
print("Odd: ", odd)
```

### Get smaller elements

```python
def small(l, x):
    s = []
    for i in l:
        if i < x:
            s.append(i)
    return s


l = [8, 100, 20, 40, 3, 7]
x = 10
print(small(l, x))
```

### Slicing(List, Tuple, String)

```python
l = [10, 20, 30, 40, 50]
print(l[0:5:2])
print(l[:4])
print(l[2:])
print(l[1:4])
print(l[4:1:-1])
print(l[-1:-6:-1])
print(l[::-1])
print(l[0:5])
print(l[:])

l1 = [10, 20, 30]
l2 = l1[:]

t1 = (10, 20, 30)
t2 = t1[:]

s1 = "geeks"
s2 = s1[:]

print(l1 is l2)
print(t1 is t2)
print(s1 is s2)
```

### Comprehension in Python

```python
def getSmaller(l, x):
    return [i for i in l if i < x]


l = [9, 15, 12, 3, 7, 11]
x = 10
print(getSmaller(l, x))
```

```python
def getEvenOdd(l):
    even = [x for x in l if x % 2 == 0]
    odd = [x for x in l if x % 2 != 0]
    return even, odd


l = [10, 3, 20, 5, 12]
even, odd = getEvenOdd(l)
print(even)
print(odd)
```

```python
s = "geekforgeeks"
l1 = [x for x in s if x in "aeiou"]
print(l1)
l2 = ["geeks", "ide", "courses", "gfg"]
l3 = [x for x in l2 if x.startswith("g")]
print(l3)
l4 = [x * 2 for x in range(6)]
print(l4)
```

```python
l1 = ["geeks", "for", "geeks", "gfg", "ide"]
l2 = [x.upper() for x in l1 if x.startswith("g")]
print(l1)
print(l2)
```

### Set Comprehensions

```python
l = [10, 20, 3, 4, 10, 20, 7, 3]
s1 = {x for x in l if x % 2 == 0}
s2 = {x for x in l if x % 2 != 0}
print(s1)
print(s2)
```

### Dictionary Comprehensions

```python
l1 = [1, 3, 4, 2, 5]
d1 = {x: x**3 for x in range(5)}
print(d1)
d2 = {x: f"ID{x}" for x in range(5)}
print(d2)
l2 = [101, 103, 102]
l3 = ["gfg", "ide", "courses"]
d3 = {l2[i]: l3[i] for i in range(len(l2))}
print(d3)
```

```python
d1 = {101: "gfg", 103: "practice", 102: "ide"}
d2 = {v: k for (k, v) in d1.items()}
print(d2)
```

### Largest element in a list

```python
def getLargest(l):
    big = l[0]
    for i in l:
        if i > big:
            big = i
    return big


l = [10, 5, 20, 8]
print(getLargest(l))
```

```python
def getLargest(l):
    for x in l:
        for y in l:
            if y > x:
                break
        else:
            return x
    return None


l = [10, 5, 20, 8]
print(getLargest(l))
```
