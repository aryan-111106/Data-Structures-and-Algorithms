### Number of digits in a number

```python
def digits(n):
    count = 0
    while n > 0:
        count += 1
        n //= 10
    return count


def main():
    n = int(input("Enter a number: "))
    print(f"No. of digits: {digits(n)}")


if __name__ == "__main__":
    main()
```

### Reverse a number

```python
def reverse(n):
    sign = -1 if n < 0 else 1
    n = abs(n)
    rev = 0
    while n > 0:
        rem = n % 10
        rev = rev * 10 + rem
        n //= 10
    rev *= sign
    return rev


def main():
    n = int(input("Enter a no.: "))
    print(f"Reversed no.: {reverse(n)}")


if __name__ == "__main__":
    main()
```

### Palindrome number

```python
def palindrome(n):
    original = n
    rev = 0
    while n > 0:
        rem = n % 10
        rev = rev * 10 + rem
        n //= 10
    return rev == original


def main():
    n = int(input("Enter a no.: "))
    if palindrome(n):
        print(f"{n} is a palindrome.")
    else:
        print(f"{n} is not a palindrome.")


if __name__ == "__main__":
    main()
```

### GCD

```python
def gcd(a, b):
    # while a and b > 0:
    #     if a > b:
    #         a = a % b
    #     else:
    #         b = b % a
    # if a == 0:
    #     return b
    # return a
    while b != 0:
        temp = b
        b = a % b
        a = temp
    return a


def main():
    a, b = map(int, input("Enter two no.: ").split())
    print(f"GCD of {a} & {b} = {gcd(a, b)}")


if __name__ == "__main__":
    main()
```

### Armstrong number

```python
def armstrong(n):
    num = n
    result = 0
    while n > 0:
        rem = n % 10
        result += rem**3
        n //= 10
    return result == num


def main():
    n = int(input("Enter a no.: "))
    if armstrong(n):
        print(f"{n} is an armstrong no.")
    else:
        print(f"{n} is not an armstrong no.")


if __name__ == "__main__":
    main()
```

### Print all divisor

```python
import math


def div(n):
    res = []
    for i in range(1, int(math.sqrt(n)) + 1):
        if n % i == 0:
            res.append(i)
            if i != n // i:
                res.append(n // i)
    return res


def main():
    n = int(input("Enter a no.: "))
    print(f"Divisors of {n} are: {div(n)}")


if __name__ == "__main__":
    main()
```

### Prime

```python
def prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


def main():
    n = int(input("Enter a no.: "))
    if prime(n):
        print(f"{n} is a prime no.")
    else:
        print(f"{n} is not a prime no.")


if __name__ == "__main__":
    main()
```
